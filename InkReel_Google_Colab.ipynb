### 3️⃣ Download Latest Rendered Project ZIP

```python
# @title 📥 Zip & Download Latest Project Outputs
import glob, os, shutil
from google.colab import files

projects = sorted(glob.glob("/content/InkReel/projects/*/"), key=os.path.getmtime, reverse=True)
if projects:
    latest = projects[0]
    proj_name = os.path.basename(os.path.normpath(latest))
    zip_out = f"/content/{proj_name}"
    archive = shutil.make_archive(zip_out, "zip", latest)
    print(f"📦 Downloading project archive: {archive}...")
    files.download(archive)
else:
    print("⚠️ No projects found in /content/InkReel/projects/ yet. Generate a video in the Web UI first!")
```
